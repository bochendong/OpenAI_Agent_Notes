# Section Agent Tests

Test the section generator and its integration with the other agents.

````python
import sys
import os
from datetime import datetime
sys.path.append(os.path.join(os.path.dirname(os.getcwd())))

from agents import Runner
from sectionAgent import section_agent
from model import Section


def section_to_markdown(section: Section) -> str:
    """Convert a Section to Markdown."""
    md = []
    
    # Title
    md.append(f"# {section.section_title}\n")
    
    # Introduction
    md.append(f"## Introduction\n\n{section.introduction}\n")
    
    # Definition
    md.append(f"## Definition\n\n{section.definition}\n")
    
    # Theorems
    if section.theorems:
        md.append(f"## Theorems\n\n")
        for i, theorem in enumerate(section.theorems, 1):
            md.append(f"### Theorem {i}\n\n")
            md.append(f"{theorem.theorem}\n\n")
            if theorem.proof:
                md.append(f"**Proof:**\n\n{theorem.proof}\n\n")
    
    # Exercises
    if section.examples:
        md.append(f"## Exercises\n\n")
        for i, exercise in enumerate(section.examples, 1):
            md.append(f"### Question {i}\n\n")
            md.append(f"**Question:** {exercise.question}\n\n")
            
            if hasattr(exercise, 'options'):
                md.append("**Options:**\n")
                for option in exercise.options:
                    md.append(f"- {option}\n")
                md.append(f"\n**Correct answer:** {exercise.correct_answer}\n\n")
            
            if hasattr(exercise, 'answer'):
                md.append(f"**Answer:** {exercise.answer}\n\n")
            
            if hasattr(exercise, 'blanks'):
                md.append("**Answer:**\n")
                for blank, answer in exercise.blanks.items():
                    md.append(f"- {blank}: {answer}\n")
                md.append("\n")
            
            if hasattr(exercise, 'proof'):
                md.append(f"**Proof:**\n\n{exercise.proof}\n\n")
            
            if hasattr(exercise, 'code_answer'):
                md.append(f"**Code solution:**\n\n```python\n{exercise.code_answer}\n```\n\n")
            
            if hasattr(exercise, 'explanation') and exercise.explanation:
                md.append(f"**Explanation:** {exercise.explanation}\n\n")
    
    # Summary
    md.append(f"## Summary\n\n{section.summary}\n")
    
    return "".join(md)


def save_section_to_markdown(section: Section, filename: str = None):
    """Save a Section as a Markdown file."""
    # Create the output directory relative to the notebook.
    # Find the Test directory from the current working directory.
    current_dir = os.getcwd()
    # Use the current directory if it is named Test. Otherwise, look in its parents.
    if os.path.basename(current_dir) == 'Test':
        test_dir = current_dir
    else:
        # Look for Test in the parent directories.
        parent = os.path.dirname(current_dir)
        if os.path.basename(parent) == 'Test':
            test_dir = parent
        else:
            # Fall back to a relative path, assuming the notebook is running in Test.
            test_dir = current_dir
    
    output_dir = os.path.join(test_dir, "output")
    output_dir = os.path.abspath(output_dir)  # Resolve the absolute path.
    os.makedirs(output_dir, exist_ok=True)
    
    # Generate the filename.
    if filename is None:
        # Use the section title as the filename after removing special characters.
        safe_title = "".join(c if c.isalnum() or c in (' ', '-', '_') else '_' for c in section.section_title)
        safe_title = safe_title.replace(' ', '_')[:50]  # Limit the filename length.
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        filename = f"{safe_title}_{timestamp}.md"
    
    # Ensure the filename ends in .md.
    if not filename.endswith('.md'):
        filename += '.md'
    
    # Full path
    filepath = os.path.join(output_dir, filename)
    
    # Convert to Markdown and save.
    markdown_content = section_to_markdown(section)
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(markdown_content)
    
    print(f"✅ Saved the section to: {filepath}")
    return filepath
````

## Test 1: Generate a full mathematics section

````python
response = await Runner.run(section_agent, "Create a section on the domain of a function with a definition, relevant theorems, three to five exercises of increasing difficulty, and a summary.")
````

````python
section = response.final_output

save_section_to_markdown(section, "function_domain.md")
````
