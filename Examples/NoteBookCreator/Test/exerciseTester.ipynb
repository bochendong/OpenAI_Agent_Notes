## Test 7: Sociology exercises


```python
response = await Runner.run(exercise_agent, "Generate three to five exercises on socialization in increasing order of difficulty. Include multiple-choice, fill-in-the-blank, and short-answer questions.")
```

```python
print(f"Generated {len(response.final_output.exercises)} questions\n")
for i, exercise in enumerate(response.final_output.exercises, 1):
    print(f"=" * 80)
    print(f"Question {i} - {type(exercise).__name__}")
    print(f"=" * 80)
    print(f"Question: {exercise.question}")
    if hasattr(exercise, 'options'):
        print(f"Options: {exercise.options}")
        print(f"Correct answer: {exercise.correct_answer}")
    if hasattr(exercise, 'answer'):
        print(f"Answer: {exercise.answer}")
    if hasattr(exercise, 'blanks'):
        print(f"Answer: {exercise.blanks}")
    if hasattr(exercise, 'explanation') and exercise.explanation:
        print(f"Explanation: {exercise.explanation}")
    print()
```
